## Ćwiczenie 3: Operacje wektorowe - analiza biznesowa
### Cel
Użyj operacji NumPy do rozwiązania problemów biznesowych. Samodzielna praca.

### Dane

```python
nazwy = ['Laptop', 'Tablet', 'Smartfon', 'Słuchawki', 'Ładowarka',
         'Klawiatura', 'Mysz', 'Monitor', 'Kamera', 'Drukarka']

ceny = np.array([3499, 1899, 2499, 349, 89, 249, 129, 1499, 599, 899])
ilosci_magazyn = np.array([15, 42, 78, 156, 312, 89, 201, 23, 45, 34])
oceny_klientow = np.array([4.7, 4.2, 4.5, 4.8, 3.9, 4.1, 4.3, 4.6, 3.8, 3.5])
```

### Zadania

**Zadanie 1:** Podstawowe statystyki cenowe

```python
print(f"Średnia cena: {ceny.mean()} zł")
print(f"Najdroższy produkt: {ceny.max()} zł")
print(f"Najtańszy produkt: {ceny.min()} zł")
print(f"Odchylenie standardowe: {ceny.std():.2f} zł")

```

```python
nazwy = np.array(nazwy)
print(f"Najdroższy produkt: {nazwy[ceny.argmax()]}")
print(f"Najtańszy produkt: {nazwy[ceny.argmin()]}")
```

**Zadanie 2:** Wartość magazynu

```python
wartosc_produktow = ceny * ilosci_magazyn
for n, wp in zip(nazwy, wartosc_produktow):
    print(f"{n} łączna wartość na magazynie: {wp} zł")
print(f"Wartość magazynu: {np.sum(ceny * ilosci_magazyn)} zł")
```

**Zadanie 3:** Analiza cenowa

#### Ceny po rabacie 15% (promocja świąteczna)

```python
promocja_swiateczna = ceny * 0.85

for nazwa, cena in zip(nazwy, promocja_swiateczna):
    print(f"{nazwa}: {cena:.2f} zł")
```

#### Ceny z VAT 23%

```python
ceny_z_vat = ceny * 1.23
for nazwa, cena in zip(nazwy, ceny_z_vat):
    print(f"{nazwa}: {cena:.2f} zł")
```

#### Ile produktów kosztuje powyżej 1000 zł?

```python
#licznik = 0
#for cena in ceny:
#    if cena > 1000:
#        licznik +=1
#    else: 
#        continue

#print(f"Ilość produktów powyżej 1000 zł: {licznik}")
print(f"Ilość produktów powyżej 1000 zł: {(ceny > 1000).sum()}")
```

**Zadanie 4:** Filtrowanie

#### Produkty z oceną >= 4.5 — nazwy i ceny

```python
nazwy = np.array(nazwy)
maska = oceny_klientow >= 4.5
nazwy_wybrane = nazwy[maska]
ceny_wybrane = ceny[maska]
for n, c in zip(nazwy_wybrane, ceny_wybrane):
    print(f"{n}: {c} zł")
```

#### Produkty poniżej 200 zł — nazwy i ilości w magazynie

```python
nazwy = np.array(nazwy)
maska = ceny < 200
ceny_ponizej = ceny[maska]
nazwy_ponizej = nazwy[maska]
ilosc_ponizej = ilosci_magazyn[maska]
for c, n, i in zip(ceny_ponizej, nazwy_ponizej, ilosc_ponizej):
    print(f"{n}: {c} zł, {i} sztuk")
```

**Zadanie 5:** Analiza zapasów

#### Produkty z niskim stanem magazynowym (< 30 szt.) — nazwy

```python
maska = ilosci_magazyn < 30
nazwy = np.array(nazwy)
produkty_niski_stan = nazwy[maska]
ilosc_na_stanie_niski_stan = ilosci_magazyn[maska]
print("Produkty z niskim stanem magazynowym:")
for p, i in zip(produkty_niski_stan, ilosc_na_stanie_niski_stan):
    print(f"- {p} x {i}")
```

#### Łączna wartość tych produktów

```python
ceny_niski_stan = ceny[maska]
#print(ceny_niski_stan)
print(f"Łączna wartość produktów, \nktórych stan magazynowych jest niski: \n {(ilosc_na_stanie_niski_stan * ceny_niski_stan).sum()}")
```

#### Ile % całego magazynu stanowią?

```python
caly_magazyn = (ceny * ilosci_magazyn).sum()
#print(caly_magazyn)
procent_magazynu = ((ilosc_na_stanie_niski_stan * ceny_niski_stan).sum()/caly_magazyn)*100
print(f"ile % warości całego magazynu: {procent_magazynu:.2f}%")
```

## Ćwiczenie 4: Benchmark — lista vs NumPy (15 min) 

```python
import time
# Przygotowanie danych
rozmiar = 1_000_000
lista = list(range(rozmiar))
tablica = np.array(lista)
```

```python
# Test 1: Pomnóż przez 2
5
start = time.perf_counter()
wynik_lista = [x * 2 for x in lista]
czas_lista = time.perf_counter()- start
start = time.perf_counter()
wynik_numpy = tablica * 2
czas_numpy = time.perf_counter()- start
print(f"=== Mnożenie ×2 ({rozmiar:,} elementów) ===")
print(f"Lista: {czas_lista*1000:.1f} ms")
print(f"NumPy: {czas_numpy*1000:.1f} ms")
print(f"NumPy jest {czas_lista/czas_numpy:.0f}× szybszy!")
```

```python
# Test 2: Suma
start = time.perf_counter()
suma_lista = sum(lista)
czas_lista = time.perf_counter()- start
start = time.perf_counter()
suma_numpy = tablica.sum()
czas_numpy = time.perf_counter()- start
print(f"\n=== Suma ({rozmiar:,} elementów) ===")
print(f"Lista: {czas_lista*1000:.1f} ms")
print(f"NumPy: {czas_numpy*1000:.1f} ms")
print(f"NumPy jest {czas_lista/czas_numpy:.0f}× szybszy!")
```

```python
# Test 3: Filtrowanie (elementy > 500000)
start = time.perf_counter()
filtr_lista = [x for x in lista if x > 500000]
czas_lista = time.perf_counter()- start
start = time.perf_counter()
filtr_numpy = tablica[tablica > 500000]
czas_numpy = time.perf_counter()- start
print(f"\n=== Filtrowanie > 500k ({rozmiar:,} elementów) ===")
print(f"Lista: {czas_lista*1000:.1f} ms")
print(f"NumPy: {czas_numpy*1000:.1f} ms")
print(f"NumPy jest {czas_lista/czas_numpy:.0f}× szybszy!")
```

## Wnioski z benchmarku
- NumPy jest **22×** szybszy w mnożeniu
- NumPy jest **9×** szybszy w sumowaniu
- NumPy jest **21×** szybszy w filtrowaniu
