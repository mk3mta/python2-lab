## Ćwiczenie 1: Series — tworzenie i operacje

```python
import pandas as pd
import numpy as np

# Series z listy — domyślny indeks (0, 1, 2, ...)
oceny = pd.Series([4.5, 3.0, 5.0, 3.5, 4.0])
print(oceny.to_string())
```

### Zadanie 1a: Utwórz Series pensje z dict:

```python
# Klucze: 'Anna', 'Jan', 'Ewa', 'Marek', 'Kasia'
# Wartości: 5500, 7200, 4800, 9100, 6300
pensje = pd.Series({'Anna' : 5500,
                    'Jan' : 7200,
                    'Ewa': 4800,
                    'Marek': 9100,
                    'Kasia': 6300})
```

### Zadanie 1b: Odpowiedz na pytania o pensje:

```python
# 1. Wyświetl imię osoby zarabiającej najwięcej (użyj idxmax())
osoba_zarobek_max = pensje.idxmax()
print(f"Osoba zarabiająca najwięcej: {osoba_zarobek_max}\n")

# 2. Oblicz średnią pensję
srednia_pensja = pensje.mean()
print(f"Średnia pensja: {srednia_pensja} zł\n")

# 3. Wyświetl imiona osób zarabiających powyżej średniej (filtrowanie boolean)
maska = pensje > srednia_pensja
print(f"Osoby zarabiające powyżej średniej: {', '.join(pensje[maska].index)}\n")

# 4. Policz ile osób zarabia poniżej 6000 (użyj len() lub .sum())
maska = pensje.values < 6000
print(f"Osoby zarabiające poniżej 6000 zł: {len(pensje[maska])}")
```

### Zadanie 1c: Oblicz pensje po zmianach:

```python
# 1. Oblicz pensje po podwyżce 10% dla wszystkich (pensje * 1.10)
podwyzka_10 = pensje * 1.1

print(f"--- Pensje po podwyżce 10% ---\n{podwyzka_10.to_string()}")
# 2. Oblicz pensje po premii 500 zł dla wszystkich (pensje + 500)
pensje_po_premii = pensje + 500
print(f"\n--- Pensje po premii 500 zł ---\n{pensje_po_premii.to_string()}")

# 3. Oblicz nową pensję = (pensja * 1.10) + 500
nowe_pensje = (pensje * 1.10) + 500
print(f"\n--- Nowe pensje (+ 10% + 500 zł) ---\n{nowe_pensje.to_string()}")
# 4. Wyświetl ranking nowych pensji od najwyższej (użyj sort_values)

print(f"--- Ranking  --- \n{nowe_pensje.sort_values().to_string()}")
```

## Zadanie 3c: Rozkłady wartości
